```python
import pandas as pd
import numpy as np
import yfinance as yf

print("pandas:", pd.__version__)
print("numpy:", np.__version__)
print("yfinance:", yf.__version__)
```

```python
ticker = "^GSPC"

sp500 = yf.download(
    ticker,
    start="2015-01-01",
    end="2026-09-01",
    auto_adjust=False
)

sp500.head()
```

```python
print("Размер данных:", sp500.shape)
print("Дата начала:", sp500.index.min())
print("Дата окончания:", sp500.index.max())
```

```python
sp500.columns = sp500.columns.get_level_values(0)

print(sp500.columns)
```
